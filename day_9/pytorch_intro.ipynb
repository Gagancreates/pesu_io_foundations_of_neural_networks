```python
# Creating tensors in different ways

# Empty tensor (uninitialized)
x = torch.empty(3, 4)
print("Empty tensor (3x4):")
print(x)

# Tensor with zeros
zeros = torch.zeros(2, 3)
print("\nZeros tensor:")
print(zeros)

# Tensor with ones
ones = torch.ones(2, 3)
print("\nOnes tensor:")
print(ones)

# Random tensor (values between 0 and 1)
random_tensor = torch.rand(2, 3)
print("\nRandom tensor:")
print(random_tensor)
```

## 4. Tensor Operations

```python
# Basic operations
x = torch.tensor([[1, 2], [3, 4]])
y = torch.tensor([[5, 6], [7, 8]])

# Element-wise addition
print("Addition:")
print(x + y)
print(torch.add(x, y))

# Element-wise multiplication
print("\nMultiplication:")
print(x * y)
print(torch.mul(x, y))

# Scalar operations
print("\nScalar addition:")
print(x + 2)

print("\nScalar multiplication:")
print(x * 2)
```

```python
# Matrix multiplication
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6], [7, 8]])

# Using matmul
result1 = torch.matmul(a, b)
print("Matrix multiplication (matmul):")
print(result1)

# Using @ operator
result2 = a @ b
print("\nMatrix multiplication (@):")
print(result2)

# Transpose
print("\nTranspose of a:")
print(a.T)
```

```python
# Squeeze and unsqueeze
x = torch.rand(1, 3, 4)
print(f"Original shape: {x.shape}")

# Remove dimensions of size 1
squeezed = torch.squeeze(x)
print(f"After squeeze: {squeezed.shape}")

# Add dimension of size 1
unsqueezed = torch.unsqueeze(squeezed, dim=0)
print(f"After unsqueeze at dim=0: {unsqueezed.shape}")
```

## 5. Autograd - Automatic Differentiation
PyTorch's autograd feature automatically computes gradients for backpropagation.

```python
# Create tensors with gradient tracking
x = torch.tensor([2.0, 3.0], requires_grad=True)
print(f"x: {x}")


y = x ** 2
z = y.sum()

print(f"y = x^2: {y}")
print(f"z = sum(y): {z}")

# Compute gradients
z.backward()
print(f"Gradient dz/dx: {x.grad}")
```

```python
# Example: Computing gradients for a simple function
x = torch.tensor(3.0, requires_grad=True)

# f(x) = 2x^2 + 3x + 1
y = 2 * x**2 + 3 * x + 1

print(f"f(3) = {y.item()}")

# Compute gradient: f'(x) = 4x + 3
y.backward()
print(f"f'(3) = {x.grad}")
print(f"Expected: {4*3 + 3}")
```

## 6. Working with Datasets

```python
from torch.utils.data import Dataset, DataLoader

# Create a simple custom dataset
class SimpleDataset(Dataset):
    def __init__(self, size=100):
        self.x = torch.randn(size, 10)
        self.y = torch.randint(0, 2, (size,))
    
    def __len__(self):
        return len(self.x)
    
    def __getitem__(self, idx):
        return self.x[idx], self.y[idx]


dataset = SimpleDataset(size=100)
dataloader = DataLoader(dataset, batch_size=16, shuffle=True)

print(f"Dataset size: {len(dataset)}")
print(f"Number of batches: {len(dataloader)}")


for batch_x, batch_y in dataloader:
    print(f"Batch X shape: {batch_x.shape}")
    print(f"Batch Y shape: {batch_y.shape}")
    break
```

## 7. Building Neural Networks
Neural networks are built using `torch.nn.Module` as the base class.

```python
# Complete MNIST FCNN Training in a Single Cell

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

# Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Define the FCNN model
class FCNN(nn.Module):
    def __init__(self):
        super(FCNN, self).__init__()
        self.fc1 = nn.Linear(28*28, 512)
        self.fc2 = nn.Linear(512, 256)
        self.fc3 = nn.Linear(256, 128)
        self.fc4 = nn.Linear(128, 10)
        
    def forward(self, x):
        x = x.view(-1, 28*28)  # Flatten
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        x = torch.relu(self.fc3(x))
        x = self.fc4(x)
        return x

# Load MNIST dataset
transform = transforms.Compose([transforms.ToTensor(), transforms.Normalize((0.1307,), (0.3081,))])
train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
test_dataset = datasets.MNIST(root='./data', train=False, download=True, transform=transform)
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=1000, shuffle=False)

# Initialize model, loss, and optimizer
model = FCNN().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Training loop
num_epochs = 5
for epoch in range(num_epochs):
    model.train()
    train_loss, correct, total = 0, 0, 0
    
    for batch_idx, (data, target) in enumerate(train_loader):
        data, target = data.to(device), target.to(device)
        optimizer.zero_grad()
        output = model(data)
        loss = criterion(output, target)
        loss.backward()
        optimizer.step()
        
        train_loss += loss.item()
        _, predicted = torch.max(output.data, 1)
        total += target.size(0)
        correct += (predicted == target).sum().item()
    
    # Evaluation
    model.eval()
    test_loss, test_correct, test_total = 0, 0, 0
    with torch.no_grad():
        for data, target in test_loader:
            data, target = data.to(device), target.to(device)
            output = model(data)
            test_loss += criterion(output, target).item()
            _, predicted = torch.max(output.data, 1)
            test_total += target.size(0)
            test_correct += (predicted == target).sum().item()
    
    print(f'Epoch [{epoch+1}/{num_epochs}]')
    print(f'  Train Loss: {train_loss/len(train_loader):.4f}, Train Acc: {100*correct/total:.2f}%')
    print(f'  Test Loss: {test_loss/len(test_loader):.4f}, Test Acc: {100*test_correct/test_total:.2f}%')

print('\nTraining complete!')
```
